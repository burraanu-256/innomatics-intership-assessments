# 1.Smart Parking Lot Management System

```python
def smart_parking(capacity, logs):
    parked = 0
    peak = 0
    
    for log in logs:
        if log == "IN":
            parked += 1
        elif log == "OUT" and parked > 0:
            parked -= 1
        
        peak = max(peak, parked)
    
    print("Currently Parked Vehicles:", parked)
    print("Peak Parking Usage:", peak)
    
    if parked > capacity:
        print("Parking Status: Exceeded Capacity")
    else:
        print("Parking Status: Available")

# Sample Input
smart_parking(50, ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"])
```

# 2.Online Food Delivery Time Estimator

```python
def estimate_delivery(distance, traffic, weather):
    time = distance * 5   # base time (5 min per km)
    
    if traffic == "High":
        time += 15
    elif traffic == "Medium":
        time += 8
    
    if weather == "Rainy":
        time += 10
    elif weather == "Storm":
        time += 20
    
    print("Estimated Delivery Time:", time, "minutes")

# Sample Input
estimate_delivery(8, "High", "Rainy")
```

# 3.Movie Theatre Seat Occupancy Analyzer

```python
def seat_occupancy(total_seats, booked_seats):
    booked = len(booked_seats)
    occupancy = (booked / total_seats) * 100
    
    print("Occupancy:", int(occupancy), "%")
    
    if occupancy == 100:
        print("Show Status: Housefull")
    elif occupancy >= 75:
        print("Show Status: Almost Full")
    else:
        print("Show Status: Seats Available")

# Sample Input
seat_occupancy(200, [1]*150)
```

# 4.Cloud Server Load Classification System

```python
def server_load(cpu_readings):
    avg = sum(cpu_readings) / len(cpu_readings)
    
    print("Average CPU Load:", int(avg), "%")
    
    if avg < 50:
        print("Server Status: Normal")
    elif avg <= 80:
        print("Server Status: Warning")
    else:
        print("Server Status: Critical")

# Sample Input
server_load([45, 60, 70, 85, 90])
```

# 5.Smart Classroom Resource Usage Monitor

```python
def classroom_usage(resources):
    overused = []
    
    for resource, hours in resources.items():
        if hours > 8:
            overused.append(resource)
    
    print("Overused Resources:", ", ".join(overused) if overused else "None")
    print("Energy Alert:", "Yes" if overused else "No")

# Sample Input
classroom_usage({
    "Projector": 6,
    "AC": 9,
    "Lights": 4
})
```

# 6.Online Event Registration Capacity Controller

```python
def event_registration(capacity, registrations):
    if registrations > capacity:
        confirmed = capacity
        waitlist = registrations - capacity
        status = "Closed"
    else:
        confirmed = registrations
        waitlist = 0
        status = "Open"
    
    print("Confirmed Registrations:", confirmed)
    print("Waitlisted Users:", waitlist)
    print("Registration Status:", status)

# Sample Input
event_registration(100, 105)
```
