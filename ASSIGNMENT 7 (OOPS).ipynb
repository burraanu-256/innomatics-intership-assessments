# 1.Smart Light Controller

```python
class SmartLight:
    def __init__(self,name):
        self.name=name
        self.status="OFF"
    def turn_on(self):
        self.status= "ON"
    def turn_off(self):
        self.status="OFF"
    def display_status(self):
        print(f"{self.name} is {self.status}")
#create a object
light=SmartLight("Bedroom Light")
light.turn_on()
light.display_status()
```

# 2.Employee ID Card System

```python
class Employee:
    def __init__(self,name,emp_id,department):
        self.name=name
        self.emp_id=emp_id
        self.department=department
    def display_id_card(self):
        print("Employee ID Card")
        print("Name:",self.name)
        print("ID:",self.emp_id)
        print("Department:",self.department)
#create a object
emp=Employee("Rahul","EMP102","IT")
emp.display_id_card()
```

# 3.Mobile Contact Record

```python
class Contact:
    def __init__(self,name,phone):
        self.name=name
        self.phone=phone
    def display_contact(self):
        print("Contact Saved")
        print("Name:",self.name)
        print("Phone:",self.phone)
# create a object
a=Contact("Anita",9876543210)
a.display_contact()
```

# 4.Product Price Tag Generator

```python
class Product:
    def __init__(self,name,price):
        self.name=name
        self.price=price
    def display_price_tag(self):
        print("Product:",self.name)
        print("price: ₹",self.price)
#create a object
p=Product("Headphones",2499)
p.display_price_tag()
```

# 5.Movie Rating Display System

```python
class Movie:
    def __init__(self,name,rating):
        self.name=name
        self.rating=rating
    def display_details(self):
        print("Movie:",self.name)
        print("Rating:",self.rating,"/ 5")
#create a object
m=Movie("Inception",4.8)
m.display_details()
```

# 6.Delivery Address Manager

```python
class Delivery:
    def __init__(self,customer_name,address):
        self.customer_name=customer_name
        self.address=address
    def display_delivery(self):
        print("Delivery Details")
        print("Customer:",self.customer_name)
        print("Address:",self.address)
#create a object
d=Delivery("Suman","Hydera)
d.display_delivery()
```
